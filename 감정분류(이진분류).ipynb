# 예제 4: 감정 분류 (이진 분류)

이 노트북은 영화 리뷰 문장을 기반으로 긍정/부정을 분류하는 머신러닝 실습입니다.

**목표**: 간단한 문장을 입력받아 긍정(1) 또는 부정(0)으로 분류하는 모델 만들기

**사용 기술**: CountVectorizer, Naive Bayes, scikit-learn

```python
# 필요한 라이브러리 불러오기
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
import pandas as pd
```

## 1. 데이터 준비하기
간단한 감정 문장 데이터셋을 수동으로 생성합니다.

```python
# 예시 데이터
texts = [
    '이 영화 정말 재밌어요', '최고의 영화였어요', '감동적이고 훌륭했어요',
    '별로였어요', '지루하고 재미없어요', '최악이에요',
    '정말 기대 이상이에요', '다시는 보고 싶지 않아요',
    '배우 연기가 훌륭해요', '스토리가 너무 평범해요'
]
labels = [1, 1, 1, 0, 0, 0, 1, 0, 1, 0]  # 1=긍정, 0=부정
```

## 2. 벡터화: 텍스트를 숫자로 변환

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)
y = labels
```

## 3. 모델 훈련

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
model = MultinomialNB()
model.fit(X_train, y_train)
```

## 4. 성능 평가

```python
y_pred = model.predict(X_test)
print("정확도:", accuracy_score(y_test, y_pred))
print("혼동 행렬:")
print(confusion_matrix(y_test, y_pred))
```

## 5. 새 문장 테스트
직접 문장을 입력해서 긍정/부정을 예측해봅니다.

```python
# 테스트 문장
test_sentences = ['정말 최고에요', '별로 안 좋았어요', '배우가 멋있었어요']
test_vectors = vectorizer.transform(test_sentences)
predictions = model.predict(test_vectors)

for sentence, pred in zip(test_sentences, predictions):
    print(f"'{sentence}' => {'긍정' if pred == 1 else '부정'}")
```
