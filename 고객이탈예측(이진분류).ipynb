# 예제 5: 고객 이탈 예측 (Churn Prediction)

이 노트북은 고객의 행동 데이터를 바탕으로 이탈(1) 또는 유지(0)를 예측하는 머신러닝 실습입니다.

**목표**: 고객 정보를 기반으로 서비스 해지 여부를 예측하는 분류 모델을 학습

**사용 기술**: Decision Tree, scikit-learn

```python
# 필요한 라이브러리 불러오기
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
import pandas as pd
```

## 1. 데이터 준비하기
샘플 고객 데이터를 생성합니다.

```python
# 예시 데이터
data = {
    'age': [25, 45, 33, 29, 60, 52, 40, 36, 50, 48],
    'membership_years': [1, 5, 2, 1, 10, 6, 3, 4, 7, 8],
    'monthly_usage': [5, 20, 7, 4, 30, 22, 10, 12, 25, 24],
    'churn': [1, 0, 1, 1, 0, 0, 0, 0, 0, 0]  # 1=이탈, 0=유지
}
df = pd.DataFrame(data)
df
```

## 2. 학습용 데이터와 테스트 데이터로 나누기

```python
X = df[['age', 'membership_years', 'monthly_usage']]
y = df['churn']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## 3. 모델 학습

```python
model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)
```

## 4. 성능 평가

```python
y_pred = model.predict(X_test)
print("정확도:", accuracy_score(y_test, y_pred))
print("혼동 행렬:")
print(confusion_matrix(y_test, y_pred))
```

## 5. 새 고객 예측하기
새로운 고객의 정보를 입력하여 이탈 여부를 예측합니다.

```python
# 새 고객 예측
new_customers = pd.DataFrame({
    'age': [30, 55],
    'membership_years': [2, 9],
    'monthly_usage': [6, 28]
})

predictions = model.predict(new_customers)
for i, result in enumerate(predictions):
    print(f"고객 {i+1} => {'이탈' if result == 1 else '유지'}")
```
